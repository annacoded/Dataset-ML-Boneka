```python
import os
import cv2
import numpy as np
from tensorflow.keras.preprocessing.image import ImageDataGenerator, img_to_array, load_img
import shutil

# 1. PERSIAPAN FOLDER
# Pastikan kamu sudah menjalankan: !unzip dataset_balance.zip -d /content/dataset_awal
dataset_dir = '/content/dataset_awal'
output_dir = '/content/dataset_hasil'

# NAMA FOLDER SUDAH DISESUAIKAN PERSIS DENGAN MILIKMU
classes = ['Boneka Domba-1', 'Boneka Anjing Laut-1']

if os.path.exists(output_dir):
    shutil.rmtree(output_dir)
os.makedirs(output_dir)

# 2. SETUP AUGMENTASI
datagen = ImageDataGenerator(
    rescale=1./255,
    rotation_range=20,
    horizontal_flip=True,
    zoom_range=0.2,
    brightness_range=[0.8, 1.2]
)

target_size = (224, 224)
augment_multiplier = 4 # Menggandakan data 4x lipat

print("--- PROSES PREPROCESSING & AUGMENTASI ---")
for cls in classes:
    # Memastikan komputer mencari di folder yang benar
    cls_input_dir = os.path.join(dataset_dir, cls)
    cls_output_dir = os.path.join(output_dir, cls)
    os.makedirs(cls_output_dir, exist_ok=True)

    # Cek apakah folder benar-benar ada
    if not os.path.exists(cls_input_dir):
        print(f"❌ ERROR: Folder '{cls}' tidak ditemukan di {dataset_dir}!")
        print("Pastikan file ZIP diekstrak dengan benar tanpa membuat folder ganda.")
        continue

    images = [f for f in os.listdir(cls_input_dir) if f.endswith(('.jpg', '.jpeg', '.png'))]
    print(f"⏳ Kelas {cls} - Jumlah Asli: {len(images)}")

    count_hasil = 0
    for img_name in images:
        img_path = os.path.join(cls_input_dir, img_name)
        img = load_img(img_path, target_size=target_size, color_mode='rgb')
        x = img_to_array(img)
        x = x.reshape((1,) + x.shape)

        i = 0
        for batch in datagen.flow(x, batch_size=1, save_to_dir=cls_output_dir, save_prefix=cls.replace(" ", "_"), save_format='jpg'):
            i += 1
            count_hasil += 1
            if i >= augment_multiplier:
                break

    print(f"✅ Kelas {cls} - Selesai! Total Akhir: {count_hasil} gambar\n")

print("🎉 Mantap! Format standar (224x224, RGB, JPG, Normalisasi 0-1) telah diterapkan.")
```
